# Repair Cafe Design Notebook

## Session 1

The initial concept focuses on helping people repair broken household appliances instead of discarding them. The main difficulty is that many users do not know how to assess a problem, which fixes are safe, or where to seek local help.

### Key idea
- Create a triage experience for common appliance symptoms.
- Offer beginner-safe repair education and confidence-building steps.
- Connect users to local repair support and repair events.

### Outcome
The first app prototype is a lightweight landing page with a quick-diagnosis flow and a learning section for beginner repair skills.
